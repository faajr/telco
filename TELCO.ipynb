```python
import pandas as pd

url = "https://raw.githubusercontent.com/faajr/telco/main/dataset%20telco.xlsx"

df = pd.read_excel(url)

df.head()
```

```python
print(df.shape)
df.info()
```

```python
print(df.columns)
```

```python
df.rename(columns={'Cluster': 'Province'}, inplace=True)
```

```python
print(df['Area'].value_counts())
```

```python
df['Province'] = df['Province'].map({
    'Surabaya': 'Jawa Timur',
    'Sby': 'Jawa Timur',
    'Yogyakarta': 'DIY',
    'Jogja': 'DIY',
    'Jakrta': 'Jakarta',
    'Bandng': 'Jawa Barat',
    'Bandung': 'Jawa Barat'
})
```

```python
print(df['Area'].value_counts())
```

```python
df['Area'] = df['Area'].map({
    'SBY': 'Surabaya',
    'JKT': 'Jakarta',
    'Jogja': 'Yogyakarta',
    'Jakrta': 'Jakarta',
    'Bandng': 'Bandung'
})
```

```python
print(df['Area'].value_counts())
```

```python
print(df['Province'].value_counts())
```

```python
df['Province'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
#menambahkan kolom profit
df['Profit'] = df['Recharge_Amount'] - df['Network_Cost']
df.head()
```

```python
import numpy as np
```

```python
df['NPS_Category'] = np.where(df['NPS_Score'] < 51, 'Buruk',
np.where((df['NPS_Score'] >= 51) & (df['NPS_Score'] <= 85), 'Bagus', 'Bagus Sekali'))
df.head()
```

```python
df['Activation_Date'].max()
```

```python
df.columns
```

```python
df['Activation_Date'] = pd.to_datetime(df['Activation_Date'], format='mixed', dayfirst=True)
df.head()
```

```python
df['Activation_Date'].head()
```

```python
df['NPS_Score'].head()
```

```python
# ARPU per Area
df.groupby('Area')['Target_ARPU'].mean()
```

```python
# NPS per Area
df.groupby('Area')['NPS_Score'].mean()
```

```python
# Berdasarkan nilai NPS_score kategori < 51 buruk Churn_Flag (Y: Churn) selain itu N
df['Churn_Flag'] = np.where(df['NPS_Score'] < 51, 'Y', 'N')
```

```python
# ARPU_Gap --> selisih pendapatan aktual (Recharge_Amount) vs target (Target_ARPU)
df['ARPU_Gap'] = df['Recharge_Amount'] - df['Target_ARPU']
```

```python
# kategorikan cat_target_ARPU jika positif = mencapai (Y) dan negatif = tidak mencapai (N)
def cat_target_ARPU (x) :
  if x <= 0 :
    return 'N'
  else :
    return 'Y'
df['cat_target_ARPU'] = df['ARPU_Gap'].apply(cat_target_ARPU)
df.head()
```

```python
df.groupby('Churn_Flag')['Activation_Date'].agg(
    ['count','mean','median','min','max']
)
```

```python
pd.crosstab(
    df['Activation_Date'],
    df['Churn_Flag'],
    normalize='index'
) * 100
```
